```python
import base64
from IPython.display import HTML

def img_to_base64(path):
    with open(path, "rb") as img:
        return base64.b64encode(img.read()).decode()

img1 = img_to_base64("/content/WhatsApp Image 2026-03-01 at 11.45.21 AM.jpeg")
img2 = img_to_base64("/content/WhatsApp Image 2026-03-01 at 11.45.23 AM.jpeg")
img3 = img_to_base64("/content/WhatsApp Image 2026-03-01 at 11.45.24 AM.jpeg")

html_content = f"""
<!DOCTYPE html>
<html>
<head>
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<style>

body {{
    margin: 0;
    font-family: Arial, sans-serif;
    text-align: center;
    color: white;
    background: black;
    overflow-x: hidden;
}}

.background {{
    position: fixed;
    width: 100%;
    height: 100%;
    background-image: url("data:image/jpeg;base64,{img1}");
    background-size: cover;
    background-position: center;
    filter: blur(20px) brightness(0.6);
    z-index: -1;
}}

h1 {{
    margin-top: 40px;
    font-size: 2.8em;
}}

.slider {{
    position: relative;
    width: 80%;
    margin: 40px auto;
}}

.slide {{
    display: none;
}}

.slide img {{
    width: 100%;
    max-height: 500px;
    object-fit: cover;
    border-radius: 20px;
    box-shadow: 0 15px 40px rgba(0,0,0,0.6);
    cursor: pointer;
    transition: transform 0.4s;
}}

.slide img:hover {{
    transform: scale(1.05);
}}

.active {{
    display: block;
}}

.caption {{
    margin-top: 20px;
    font-size: 1.3em;
}}

.buttons {{
    margin-top: 40px;
    position: relative;
    height: 200px;
}}

button {{
    padding: 15px 40px;
    font-size: 18px;
    border: none;
    border-radius: 40px;
    cursor: pointer;
    margin: 10px;
}}

#yesBtn {{
    background-color: #ff1744;
    color: white;
}}

#noBtn {{
    background-color: white;
    color: #ff1744;
    position: absolute;
}}

.hidden {{
    display: none;
}}

#finalMessage {{
    font-size: 2.5em;
    margin-top: 40px;
}}

.heart {{
    position: fixed;
    bottom: -20px;
    font-size: 24px;
    animation: floatUp linear infinite;
}}

@keyframes floatUp {{
    from {{ transform: translateY(0); opacity: 1; }}
    to {{ transform: translateY(-110vh); opacity: 0; }}
}}

</style>
</head>
<body>

<div class="background"></div>

<h1>Every Picture Has A Story 💖</h1>

<div class="slider">
    <div class="slide active">
        <img src="data:image/jpeg;base64,{img1}">
        <div class="caption">This smile made me fall for you ❤️</div>
    </div>
    <div class="slide">
        <img src="data:image/jpeg;base64,{img2}">
        <div class="caption">You look beautiful in every moment ✨</div>
    </div>
    <div class="slide">
        <img src="data:image/jpeg;base64,{img3}">
        <div class="caption">And I want to create forever memories with you 💍</div>
    </div>
</div>

<div class="buttons">
    <button id="yesBtn">Yes 💖</button>
    <button id="noBtn">No 🙈</button>
</div>

<div id="finalMessage" class="hidden">
    I LOVE YOU ❤️ <br><br>
    Will you be my girlfriend?
</div>

<script>

let slideIndex = 0;
const slides = document.querySelectorAll(".slide");

function showSlides() {{
    slides.forEach(slide => slide.classList.remove("active"));
    slideIndex++;
    if (slideIndex > slides.length) slideIndex = 1;
    slides[slideIndex-1].classList.add("active");
}}

setInterval(showSlides, 3000);

const noBtn = document.getElementById("noBtn");
const yesBtn = document.getElementById("yesBtn");
const finalMessage = document.getElementById("finalMessage");

function moveNoButton() {{
    const maxX = window.innerWidth - noBtn.offsetWidth - 20;
    const maxY = window.innerHeight - noBtn.offsetHeight - 20;
    noBtn.style.left = Math.random() * maxX + "px";
    noBtn.style.top = Math.random() * maxY + "px";
}}

noBtn.addEventListener("mouseover", moveNoButton);
noBtn.addEventListener("click", moveNoButton);

yesBtn.addEventListener("click", () => {{
    document.querySelector(".buttons").classList.add("hidden");
    finalMessage.classList.remove("hidden");
    launchHearts();
}});

function launchHearts() {{
    for (let i = 0; i < 40; i++) {{
        let heart = document.createElement("div");
        heart.innerHTML = "❤️";
        heart.classList.add("heart");
        heart.style.left = Math.random() * 100 + "vw";
        heart.style.animationDuration = (Math.random() * 3 + 3) + "s";
        document.body.appendChild(heart);
    }}
}}

</script>

</body>
</html>
"""

HTML(html_content)
```
